```python
# ============================================================
# 1. A Single Gradient Update
# ============================================================
# 使用一個非常簡單的模型：
# prediction = weight × x
#
# 透過 loss.backward() 計算 gradient，
# 觀察 Gradient Descent 如何取得參數更新方向。

import torch

x = torch.tensor([2.0])
y = torch.tensor([10.0])

weight = torch.tensor([1.0], requires_grad=True)

prediction = weight * x
loss = (prediction - y) ** 2

loss.backward()

print("Prediction:", prediction.item())
print("Loss:", loss.item())
print("Gradient:", weight.grad.item())


# ============================================================
# 2. Manually Update the Weight
# ============================================================
# Gradient 告訴我們應該往哪個方向調整參數，
# Learning Rate 決定每次調整的幅度。
#
# Gradient Descent：
# weight = weight - learning_rate × gradient

learning_rate = 0.01

with torch.no_grad():
    weight -= learning_rate * weight.grad

print("Updated weight:", weight.item())


# 重新計算更新後的 prediction 和 loss

prediction = weight * x
loss = (prediction - y) ** 2

print("New prediction:", prediction.item())
print("New loss:", loss.item())


# ============================================================
# Observation
# ============================================================
# 更新 Weight 後：
#
# Weight：1.00 → 1.32
# Prediction：2.00 → 2.64
# Loss：64.00 → 54.17
#
# 只進行一次 Gradient Descent 更新，
# Loss 就下降了。
```

```python
# ============================================================
# 3. Repeated Gradient Descent
# ============================================================
# 前面只更新了一次 Weight。
# 現在讓模型重複計算 gradient 並更新 Weight，
# 觀察它是否會逐漸找到最佳參數。
#
# 目標：
# x = 2
# y = 10
#
# 因此理想的 Weight 應該接近 5，
# 因為 5 × 2 = 10。

weight = torch.tensor([1.0], requires_grad=True)
learning_rate = 0.01

for step in range(100):
    # 根據目前的 Weight 計算 prediction 和 loss
    prediction = weight * x
    loss = (prediction - y) ** 2

    # 清除上一輪的 gradient
    weight.grad = None

    # 計算目前的 gradient
    loss.backward()

    # 根據 gradient 更新 Weight
    with torch.no_grad():
        weight -= learning_rate * weight.grad

    # 每 10 次觀察一次訓練狀況
    if step % 10 == 0:
        print(
            f"Step {step}, "
            f"Weight: {weight.item():.4f}, "
            f"Loss: {loss.item():.4f}"
        )


# ============================================================
# Observation
# ============================================================
# Weight 會逐漸接近 5，
# Loss 則會逐漸接近 0。
#
# 這表示 Gradient Descent 能夠透過反覆：
#
# 1. 計算 prediction
# 2. 計算 loss
# 3. 計算 gradient
# 4. 更新 Weight
#
# 讓模型逐漸找到較好的參數。
```

```python
# ============================================================
# 4. Effect of Learning Rate
# ============================================================
# Learning Rate 決定每次參數更新的幅度。
#
# 我們比較不同 Learning Rate 對 Gradient Descent 的影響：
# 0.01 → 小步更新
# 0.1  → 較大步更新
# 0.5  → 非常大的步伐
# 1.0  → 可能造成震盪與發散

learning_rates = [0.01, 0.1, 0.5, 1.0]

for learning_rate in learning_rates:
    weight = torch.tensor([1.0], requires_grad=True)

    print(f"\nLearning Rate: {learning_rate}")

    for step in range(10):
        prediction = weight * x
        loss = (prediction - y) ** 2

        weight.grad = None
        loss.backward()

        with torch.no_grad():
            weight -= learning_rate * weight.grad

    print(f"Final Weight: {weight.item():.4f}")
    print(f"Final Loss: {loss.item():.4f}")


# ============================================================
# Observation
# ============================================================
# Learning Rate 並不是越大越好。
#
# 太小：
# → 每次更新幅度小，收斂速度慢
#
# 適當：
# → 能夠有效地往最佳解靠近
#
# 太大：
# → 可能跨過最佳解
# → 可能造成震盪
# → 如果過大，甚至可能發散
```
